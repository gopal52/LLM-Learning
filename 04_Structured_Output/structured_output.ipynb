# Pydantic

Pydantic is a Python library used to define and validate structured data using Python classes.

Just remember:
Pydantic helps us create a fixed structure for data.


Why is Pydantic Popular?

Because it gives:

Structure
Type checking
Validation
Easy integration with LangChain
