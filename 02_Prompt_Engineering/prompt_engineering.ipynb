# What is a prompt?

A prompt is simply the instruction or question you give to an AI.
Prompt = Anything you ask AI.

# What is Prompt Engineering?
The process of writing clear and effective prompts to get better responses from an AI.

Engineering means designing something carefully.
For example:
Instead of asking: Tell me about Docker.
You ask: Explain Docker for a beginner in less than 100 words with one real-world example.
The second prompt is much clearer.

# 1. Goal:
 The main task or objective that the AI should perform.

# 2. Context:
 Additional background information that helps the AI understand the request and provide a more relevant answer.

# 3. Constraints:
A constraint is simply a rule or a limitation that tells the AI how the answer should be generated.

Think of it as:
"Do the task, but follow these rules."

Prompt: Explain Docker in 100 words using 3 bullet points.
Let's divide it.
Goal: Explain Docker.
Constraints
In 100 words.
Use 3 bullet points.

These are rules the AI must follow.

# 4. Example (Optional):
 An Example means
Showing the AI one or more sample inputs and outputs so it learns the pattern you want.


# 5. Role (Optional):
 Defines the personality, profession, or perspective the AI should adopt while answering.
Prompt: You are an English grammar teacher. Correct my sentence.
The AI behaves like an English teacher.

# 1.Zero-Shot Prompting:

Asking the AI to perform a task without providing any examples.

Prompt

Write a Java program to print Fibonacci numbers.

Example given?

❌ No

Technique?

✅ Zero-Shot

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
load_dotenv()
```

```python
llm = ChatOpenAI(
    model="gpt-5.4-nano"
)
# Store the ChatGPT model inside the variable called llm.
```

```python
# Create Messages: Now we tell the AI its role and ask a question.

messages = [
    SystemMessage(
        content="You are a helpful Cricketer coach you dont know about python labgauge"
    ),

    HumanMessage(
        content="Explain Python in one line."
    )
]

#SystemMessage(...)---This is NOT the question, It tells the AI how to behave.

#HumanMessage(...)---This is the actual user question.
```

```python
response = llm.invoke(messages)
print(response.content)
```

```python
messages = [
    SystemMessage(
        content="""
You are only a cricket coach.

You ONLY answer cricket-related questions.

If anyone asks about any other topic,
reply exactly with:

Sorry, I only answer cricket questions.
"""
    ),

    HumanMessage(
        content="Explain Python in one line."
    )
]

response = llm.invoke(messages)

print(response.content)
```

# 👨‍🏫 Lecture 2: One-Shot Prompting
One-Shot Prompting means giving the AI one example before asking it to perform the actual task.

```python
from langchain_core.messages import SystemMessage, HumanMessage

messages = [

    SystemMessage(
        content="You are a helpful assistant."
    ),

    HumanMessage(
        content="""
Example:

Input:
Apple

Output:
APPLE

Now convert:

Input:
docker

Output:
"""
    )
]

response = llm.invoke(messages)

print(response.content)
```

# 👨‍🏫 Lecture 4: Chain of Thought (CoT) Prompting
Chain of Thought Prompting encourages the AI to reason through a problem step by step before producing the final answer.

When should we use CoT?

✅ Math problems

✅ Logical reasoning

✅ Business decisions

✅ Coding logic

✅ Multi-step calculations

```python
messages = [

    SystemMessage(
        content="You are a logical reasoning expert."
    ),

    HumanMessage(
        content="""
A train travels 60 km in one hour.

How far will it travel in 5 hours?

Think step by step.
"""
    )
]

response = llm.invoke(messages)

print(response.content)
```

# 👨‍🏫 Lecture 5: Persona Prompting (Role Prompting)
Persona Prompting, also called Role Prompting.

A Persona is the identity or character that the AI should act as.

For example:

👨‍🏫 Teacher
👨‍💻 DevOps Engineer
👨‍⚕️ Doctor
⚖️ Lawyer
👨‍🍳 Chef
💼 HR Interviewer

The AI is still the same model.

Only its style of answering changes.

```python
from langchain_core.messages import SystemMessage, HumanMessage

messages = [

    SystemMessage(
        content="You are an experienced DevOps Engineer."
    ),

    HumanMessage(
        content="Explain Docker in one line."
    )
]

response = llm.invoke(messages)

print(response.content)
```

# 👨‍🏫 Lecture 6: Tree of Thought (ToT) Prompting
Tree of Thought Prompting encourages the AI to generate and evaluate multiple possible reasoning paths before selecting the best answer.

Imagine a tree.

              Problem
                 │
        ┌────────┼────────┐
        │        │        │
     Idea A   Idea B   Idea C
        │        │        │
     Evaluate Evaluate Evaluate
        │        │        │
        └────────┼────────┘
                 │
          Best Solution



          Real-Life Example

# Suppose you have ₹50,000.

You want to buy a laptop.

Will you buy the first laptop you see?

No.

You compare:

Option A

Dell

Option B

HP

Option C

Lenovo

Then you compare:

Price
RAM
Battery
Processor

Finally you choose the best one.

That is Tree of Thought

```python
from langchain_core.messages import SystemMessage, HumanMessage
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage
load_dotenv()


messages = [

    SystemMessage(
        content="You are a career mentor."
    ),

    HumanMessage(
        content="""
I want to switch my career as i am working as ioS developer and learning GenAI and Agentic AI

Think of three possible career paths.

Compare each path.

Recommend the best one with reasons.
"""
    )
]

llm = ChatOpenAI(
    model="gpt-5.4-nano"
)
response = llm.invoke(messages)

print(response.content)
```
